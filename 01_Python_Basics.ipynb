### 1. Basic Concepts

## Variables

– What is a Variable?

A variable is a name used to store a value in Python.
It is like a box that keeps some information inside it.
We can store numbers, text, or other data in a variable.
The value inside the variable can be changed whenever we want.

### Rules for Creating Variables

1. A variable name must start with a letter (A-Z or a-z) or an underscore (_).

2. A variable name cannot start with a number.

3. A variable name can contain letters, numbers, and underscores (_).

4. Spaces are not allowed in variable names.

5. Variable names are case-sensitive.
   - `name` and `Name` are different variables.

6. Do not use Python keywords as variable names.
   - Examples: `if`, `for`, `class`, `while`.

### Example

My name is **Saiprakash**.

Instead of writing "Saiprakash" many times, I can store it in a variable.

Example:

name = "Saiprakash"

Now whenever I use `name`, Python will use the value "Saiprakash".

This makes the program simple and easy to update.

````python
# Creating variables

name = "Saiprakash"
age = 23
college = "ABC College"

print("My Name:", name)
print("My Age:", age)
print("My College:", college)
````

### Explanation

In this program, I created three variables:
 `name` stores my name.
 `age` stores my age.
 `college` stores my college name.

The `print()` function displays the values stored in the variables.

This shows that variables are used to store information, and we can use the variable name whenever we need that value.

The `type()` function is used to check the data type before and after type casting.

### 2. Control Flow

## if / elif / else

### What is if / elif / else?

The `if`, `elif`, and `else` statements are used to make decisions in a program.

- `if` checks the first condition.
- `elif` checks another condition if the first one is false.
- `else` runs when none of the conditions are true.

### Example

Suppose a student gets 75 marks.

If the marks are 90 or above, the grade is A.

If the marks are 75 or above, the grade is B.

Otherwise, the grade is C.

````python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
else:
    print("Grade C")
````

### Explanation

In this program, Python first checks the `if` condition.

Since it is false, it checks the `elif` condition.

The `elif` condition is true, so it prints **Grade B**.

The `else` block is skipped.

## for loop

### What is a for loop?

A `for` loop is used to repeat a block of code a fixed number of times.

It is commonly used to go through numbers, lists, strings, or other collections.

### Example

Suppose I want to print the numbers from 1 to 5.

Instead of writing `print()` five times, I can use a `for` loop.

````python
for i in range(1, 6):
    print(i)
````

### Explanation

In this program, the `for` loop runs from 1 to 5 using the `range()` function.

The value of `i` changes in each iteration, and `print(i)` displays the current number.

This avoids writing the same code multiple times.

## while loop

### What is a while loop?

A `while` loop is used to repeat a block of code as long as a condition is true.

When the condition becomes false, the loop stops.

### Example

Suppose I want to print the numbers from 1 to 5.

I can use a `while` loop that keeps running until the number becomes greater than 5.

````python
i = 1

while i <= 5:
    print(i)
    i += 1
````

### Explanation

In this program, the variable `i` starts with the value 1.

The `while` loop keeps running as long as `i` is less than or equal to 5.

After printing each number, `i` is increased by 1.

When `i` becomes 6, the condition is false, and the loop stops.

## break

### What is break?

The `break` statement is used to stop a loop immediately.

When Python finds `break`, it exits the loop and continues with the next statement after the loop.

### Example

Suppose I want to print numbers from 1 to 5.

If the number becomes 3, I want to stop the loop.

I can use the `break` statement.

````python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
````

### Explanation

In this program, the `for` loop starts printing numbers from 1.

When the value of `i` becomes 3, the `break` statement stops the loop immediately.

So, only 1 and 2 are printed.

## continue

### What is continue?

The `continue` statement is used to skip the current iteration of a loop.

When Python finds `continue`, it skips the remaining code in the current iteration and moves to the next iteration.

### Example

Suppose I want to print numbers from 1 to 5.

If the number is 3, I want to skip it and print the remaining numbers.

I can use the `continue` statement.

````python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````

### Explanation

In this program, the `for` loop prints numbers from 1 to 5.

When the value of `i` becomes 3, the `continue` statement skips that iteration.

So, the number 3 is not printed, and the loop continues with 4 and 5.

## pass

### What is pass?

The `pass` statement is a placeholder in Python.

It does nothing when the program runs.

It is used when we want to write code later but need to keep the program free from errors.

### Example

Suppose I want to create an `if` statement, but I have not written the code yet.

I can use the `pass` statement as a placeholder.

````python
age = 18

if age >= 18:
    pass

print("Program Completed")
````

### Explanation

In this program, the `if` condition is true.

The `pass` statement does nothing, so Python skips it and continues to the next line.

Finally, the program prints **Program Completed**.

Easy way to remember
break → Stops the loop.

continue → Skips the current iteration and continues with the next.

pass → Does nothing; it is used as a placeholder for future code.

A tuple is a collection used to store multiple items in a single variable.

## List Comprehension

### What is List Comprehension?

List comprehension is a short and simple way to create a list.

It allows us to create a new list using a single line of code.

It makes the code shorter and easier to read.

### Example

Suppose I want to create a list of numbers from 1 to 5.

Instead of using a `for` loop, I can use list comprehension.

````python
numbers = [i for i in range(1, 6)]

print(numbers)
````

### Explanation

In this program, list comprehension creates a list of numbers from 1 to 5.

## Dictionary Comprehension

### What is Dictionary Comprehension?

Dictionary comprehension is a short and simple way to create a dictionary.

It creates a dictionary in a single line of code.

It makes the code shorter and easier to read.

### Example

Suppose I want to create a dictionary with numbers as keys and their squares as values.

Instead of using a `for` loop, I can use dictionary comprehension.

````python
numbers = {i: i*i for i in range(1, 6)}

print(numbers)
````

### Explanation

In this program, dictionary comprehension creates a dictionary.

The numbers from 1 to 5 are used as keys.

Their squares are stored as values.

The `print()` function displays the dictionary.

### 4. Functions

## Functions

### Function Definition

#### What is a Function?

A function is a block of code that performs a specific task.

It helps us avoid writing the same code again and again.

In Python, a function is created using the `def` keyword.

### Example

Suppose I want to print a welcome message many times.

Instead of writing the same `print()` statement again and again, I can create a function.

````python
def greet():
    print("Welcome to Python")

greet()
````

### Explanation

In this program, I created a function named `greet()` using the `def` keyword.

When I call the function using `greet()`, it prints the welcome message.

### Arguments

#### What are Arguments?

Arguments are values passed to a function when it is called.

They allow the function to work with different values.

### Example

Suppose I want to greet different people.

Instead of creating a separate function for each person, I can pass the person's name as an argument.

````python
def greet(name):
    print("Hello", name)

greet("Sai")
````

### Explanation

In this program, the function `greet()` has one argument called `name`.

When I call `greet("Sai")`, the value `"Sai"` is passed to the function.

The function prints the greeting using the given name.

### Default Arguments

#### What are Default Arguments?

A default argument has a value that is used if no argument is passed.

### Example

Suppose I want to greet a user.

If no name is given, the function will use "Guest" as the default name.

````python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Sai")
````

### Explanation

The function has a default value `"Guest"`.

When no argument is passed, it prints "Hello Guest".

When an argument is passed, it uses that value instead of the default.

### Keyword Arguments

#### What are Keyword Arguments?

In keyword arguments, we pass values using the parameter name.

This makes the code easier to read.

````python
def student(name, age):
    print(name)
    print(age)

student(age=23, name="Sai")
````

### Explanation

In the function definition,

- `name` and `age` are parameter names.

In the function call,

- `name="Sai"` passes the value `"Sai"` to the `name` parameter.
- `age=23` passes the value `23` to the `age` parameter.

Since we use the parameter names, the order of the arguments does not matter.

### Variable-Length Arguments (`*args`, `**kwargs`)

#### What are Variable-Length Arguments?

Variable-length arguments allow a function to accept any number of arguments.

- `*args` is used to pass multiple positional arguments.
- `**kwargs` is used to pass multiple keyword arguments.

### Example

Suppose I want to add different numbers.

Instead of creating separate functions for 2 numbers, 3 numbers, or 4 numbers, I can use `*args`.

Similarly, if I want to pass different information like name, age, and city, I can use `**kwargs`.

*args

````python
def add(*numbers):
    print(sum(numbers))

add(10, 20)
add(10, 20, 30)
````

### Explanation

In this program, `*numbers` accepts any number of values.

The values are stored as a tuple.

The `sum()` function adds all the values and prints the result.

**kwargs

````python
def student(**details):
    print(details)

student(name="Sai", age=22)
````

### Explanation

In this program, `**details` accepts keyword arguments.

The values are stored as a dictionary.

We can access the values using their keys.

### Lambda Functions

#### What is a Lambda Function?

A lambda function is a small anonymous function.

It is created using the `lambda` keyword.

It is mainly used for short and simple operations.

### Example

Suppose I want to add two numbers.

Instead of creating a normal function, I can use a lambda function.

````python
add = lambda a, b: a + b

print(add(10, 20))
````

### Explanation

In this program, I created a lambda function to add two numbers.

The lambda function takes two arguments, `a` and `b`.

It returns the sum of the two numbers.

### Recursive Functions

#### What is a Recursive Function?

A recursive function is a function that calls itself.

It is used to solve problems by repeating the same task until a condition is met.

### Example

Suppose I want to print numbers from 1 to 5.

A recursive function can call itself until it reaches 5.

````python
def numbers(n):
    if n <= 5:
        print(n)
        numbers(n + 1)

numbers(1)
````

### Explanation

In this program, the function `numbers()` calls itself.

It prints the current number and then calls itself with the next number.

The recursion stops when the value becomes greater than 5.

### 5. Advanced Python

### Iterators

#### What is an Iterator?

An iterator is an object that allows us to access elements one by one.

It is created using the `iter()` function.

The `next()` function is used to get the next element.

### Example

Suppose I have a list of numbers.

I can use an iterator to access each number one at a time.

````python
numbers = [10, 20, 30]

it = iter(numbers)

print(next(it))
print(next(it))
print(next(it))
````

### Explanation

In this program, `iter()` creates an iterator from the list.

The `next()` function returns one element at a time.

After all elements are accessed, the iterator ends.

### Generators

#### What is a Generator?

A generator is a special type of function that produces values one at a time.

It uses the `yield` keyword instead of `return`.

### Example

Suppose I want to generate numbers one by one.

I can use a generator function.

````python
def numbers():
    yield 1
    yield 2
    yield 3

for i in numbers():
    print(i)
````

### Explanation

In this program, the function uses `yield` to generate values.

The values are produced one at a time when the function is executed.

This helps save memory because values are generated only when needed.

### yield

#### What is `yield`?

The `yield` keyword is used inside a generator function.

It returns one value at a time and remembers the function's state.

### Example

Suppose I want to return values one after another.

I can use the `yield` keyword.

````python
def fruits():
    yield "Apple"
    yield "Mango"
    yield "Orange"

for fruit in fruits():
    print(fruit)
````

### Explanation

In this program, `yield` returns one fruit at a time.

Each time the loop runs, the next value is generated.

This continues until all values have been returned.

### Decorators

#### What is a Decorator?

A decorator is a function that adds extra functionality to another function without changing its original code.

Decorators are written using the `@` symbol.

### Example

Suppose I want to display a message before calling a function.

I can use a decorator.

````python
def decorator(func):
    def message():
        print("Welcome")
        func()
    return message

@decorator
def greet():
    print("Hello")

greet()
````

### Explanation

In this program, `decorator()` adds an extra message before calling the `greet()` function.

The `@decorator` symbol applies the decorator to the function.

This allows us to add new behavior without modifying the original function.

### Closures

#### What is a Closure?

A closure is a function inside another function.

The inner function can use the variables of the outer function even after the outer function has finished.

### Example

Suppose I want to remember a person's name.

I can use a closure.

````python
def outer(name):
    def inner():
        print("Hello", name)
    return inner

greet = outer("Sai")
greet()
````

### Explanation

In this program, `inner()` is inside `outer()`.

The inner function remembers the value of `name` even after `outer()` finishes.

This is called a closure.

### Global Keyword

#### What is `global`?

The `global` keyword is used to modify a variable that is created outside the function.

### Example

Suppose I want to change a variable that is created outside the function.

I can use the `global` keyword.

````python
count = 10

def update():
    global count
    count = 20

update()
print(count)
````

### Explanation

In this program, `count` is a global variable.

The `global` keyword allows the function to modify the variable.

After calling the function, the value of `count` changes from `10` to `20`.

### Nonlocal Keyword

#### What is `nonlocal`?

The `nonlocal` keyword is used to modify a variable from the outer function.

It is used only inside nested functions.

### Example

Suppose I want to change a variable from the outer function.

I can use the `nonlocal` keyword.

````python
def outer():
    count = 10

    def inner():
        nonlocal count
        count = 20

    inner()
    print(count)

outer()
````

### Explanation

In this program, `count` belongs to the `outer()` function.

The `inner()` function uses the `nonlocal` keyword to modify it.

After calling `inner()`, the value of `count` becomes `20`.

### 6. Built-in Functions

### map()

#### What is `map()`?

The `map()` function applies a function to every item in an iterable.

### Example

Suppose I want to double all the numbers in a list.

I can use the `map()` function.

````python
numbers = [1, 2, 3]

result = list(map(lambda x: x * 2, numbers))

print(result)
````

### Explanation

In this program, `map()` applies the lambda function to every number.

Each number is multiplied by 2.

The result is converted into a list.

### filter()

#### What is `filter()`?

The `filter()` function returns only the elements that satisfy a condition.

### Example

Suppose I want only even numbers from a list.

I can use the `filter()` function.

````python
numbers = [1, 2, 3, 4, 5]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
````

### Explanation

In this program, `filter()` checks each number.

Only even numbers are returned.

### zip()

#### What is `zip()`?

The `zip()` function combines two or more iterables into one.

### Example

Suppose I have names and marks.

I can combine them using `zip()`.

````python
names = ["Sai", "Ram"]
marks = [90, 85]

print(list(zip(names, marks)))
````

### Explanation

In this program, `zip()` combines the elements from both lists.

Each name is paired with its mark.
