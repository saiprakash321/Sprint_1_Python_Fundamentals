# Module 6 – SQL (Structured Query Language)

## Introduction

Structured Query Language (SQL) is a standard language used to create, manage, and manipulate relational databases. It enables users to store, retrieve, update, and analyze data efficiently using SQL commands and queries.




## 1. Database Design

### 1.1 Database

A database is an organized collection of related data that is stored electronically for efficient storage, retrieval, and management.

#### Why do We Use It?

- To store large amounts of data.
- To organize data efficiently.
- To retrieve and update data quickly.
- To reduce data redundancy.
- To maintain data consistency.

#### Example

A **School Database** stores information about students, teachers, courses, and marks.

#### Explanation

A database provides a centralized location for storing, organizing, and managing related information. It enables users and applications to access, update, and retrieve data efficiently.

#### Real-World Example

Banking systems, hospital management systems, school management systems, and e-commerce websites use databases to store, organize, and manage large amounts of information efficiently.

#### Note

A well-designed database improves performance, reduces data redundancy, and maintains data integrity.

### 1.2 Tables

A table is a structure inside a database that stores data in rows and columns. Each table contains information about a specific entity.

#### Why do We Use It?

- To organize related data in a structured format.
- To store records efficiently.
- To make searching, updating, and managing data easier.

#### Example

**Students Table**

| Student_ID | Name  | Department |
|-------------|-------|------------|
| 101 | Rahul | CSE |
| 102 | Priya | ECE |
| 103 | Ravi | ISE |

#### Explanation

Each **row** represents a single record, and each **column** represents a specific attribute of the data.

#### Note

A database can contain multiple tables, and each table stores information about a specific entity.

### 1.3 Primary Key

A Primary Key is a column that uniquely identifies each row in a table. It cannot contain duplicate or NULL values.

#### Why do We Use It?

- To uniquely identify each record.
- To prevent duplicate entries.
- To maintain data integrity.

#### Example

| Student_ID | Name |
|------------|------|
| 101 | Rahul |
| 102 | Priya |
| 103 | Ravi |

#### Explanation

Here, **Student_ID** is the Primary Key because every student has a unique ID.

#### Note

A table can have only one Primary Key, but it can consist of one or more columns (Composite Key).

### 1.4 Foreign Key

A Foreign Key is a column that links one table to another by referring to the Primary Key of another table.

#### Why do We Use It?

- To establish relationships between tables.
- To maintain referential integrity.
- To avoid duplicate data.

#### Example

**Students Table**

| Student_ID | Name |
|------------|------|
| 101 | Rahul |
| 102 | Priya |

**Marks Table**

| Mark_ID | Student_ID | Marks |
|----------|------------|-------|
| 1 | 101 | 90 |
| 2 | 102 | 85 |

#### Explanation

Here, **Student_ID** in the **Marks** table is a Foreign Key because it refers to the **Student_ID** in the **Students** table.

#### Note

A Foreign Key helps connect related tables and ensures that only valid referenced values are stored.

### 1.5 Relationships

A relationship is the connection between two or more tables using Primary Keys and Foreign Keys.

#### Types of Relationships

- **One-to-One:** One record in a table is related to one record in another table.
- **One-to-Many:** One record in a table is related to multiple records in another table.
- **Many-to-Many:** Multiple records in one table are related to multiple records in another table.

#### Why do We Use It?

- To connect related tables.
- To reduce data redundancy.
- To improve database organization.
- To maintain data consistency.

#### Example

One student can have many marks.

#### Explanation

The **Students** table is connected to the **Marks** table using **Student_ID**, creating a relationship between the two tables.

#### Real-World Example

In a school database, one student can have multiple marks, creating a one-to-many relationship between the Students and Marks tables.

### 1.6 Constraints

Constraints are rules applied to table columns to ensure the data is accurate and valid. They are defined while creating a table and inserting values.

#### Common Constraints

- **NOT NULL** – The column cannot contain NULL values.
- **UNIQUE** – All values in the column must be different.
- **PRIMARY KEY** – Uniquely identifies each row in a table.
- **FOREIGN KEY** – Establishes a relationship between two tables.
- **CHECK** – Ensures that values satisfy a specified condition.
- **DEFAULT** – Assigns a default value if no value is provided.

#### Why do We Use It?

- To maintain data integrity.
- To prevent invalid data.
- To enforce business rules.
- To improve data consistency.

#### Example

```sql
CREATE TABLE Students (
    Student_ID INT PRIMARY KEY,
    Name VARCHAR(50) NOT NULL,
    Age INT CHECK (Age >= 18),
    City VARCHAR(30) DEFAULT 'Chennai'
);
```

#### Explanation

- **PRIMARY KEY** makes `Student_ID` unique.
- **NOT NULL** ensures `Name` cannot be empty.
- **CHECK** ensures `Age` is at least 18.
- **DEFAULT** sets `City` to `"Chennai"` if no value is provided.

#### Note

Constraints help maintain data integrity by preventing invalid or inconsistent data from being stored in the database.

# 2. Queries

The **Queries** section is used to retrieve, filter, sort, group, and analyze data stored in database tables.

The main SQL queries are:

- SELECT
- WHERE
- ORDER BY
- GROUP BY
- HAVING
- DISTINCT
- LIMIT

### 2.1 SELECT

The **SELECT** statement is used to retrieve data from one or more columns of a table.

#### Why do We Use It?

- To retrieve data from a table.
- To display specific columns.
- To view stored records.

#### Syntax

```sql
SELECT column_name
FROM table_name;
```

#### Example

```sql
SELECT Name, Age
FROM Students;
```

#### Explanation

This query displays the **Name** and **Age** columns from the **Students** table.

#### Note

The `SELECT *` statement retrieves all columns from a table.

### 2.2 WHERE

The **WHERE** clause is used to filter rows based on a specified condition.

#### Why do We Use It?

- To retrieve specific records from a table.
- To filter data based on one or more conditions.
- To reduce unwanted results.

#### Syntax

```sql
SELECT column_name
FROM table_name
WHERE condition;
```

#### Example

```sql
SELECT *
FROM Students
WHERE Age > 20;
```

#### Explanation

This query displays only the students whose **Age** is greater than **20**.

#### Note

The **WHERE** clause supports comparison operators (`=`, `>`, `<`, `>=`, `<=`, `!=`) and logical operators (`AND`, `OR`, `NOT`).

### 2.3 ORDER BY

The **ORDER BY** clause is used to sort the query results in ascending (`ASC`) or descending (`DESC`) order.

#### Why do We Use It?

- To sort records in a specific order.
- To improve readability of query results.
- To arrange data based on one or more columns.

#### Syntax

```sql
SELECT column_name
FROM table_name
ORDER BY column_name ASC;
```

#### Example

```sql
SELECT *
FROM Students
ORDER BY Age ASC;
```

#### Explanation

This query sorts the students by **Age** from the smallest to the largest.

#### Note

`ASC` is the default sorting order. Use `DESC` to sort in descending order.

### 2.4 GROUP BY

The **GROUP BY** clause is used to group rows that have the same values in a specified column.

#### Why do We Use It?

- To group similar records.
- To perform aggregate calculations.
- To summarize data.

#### Syntax

```sql
SELECT column_name, aggregate_function(column_name)
FROM table_name
GROUP BY column_name;
```

#### Example

```sql
SELECT Department, COUNT(*)
FROM Employees
GROUP BY Department;
```

#### Explanation

This query displays the total number of employees in each department.

#### Note

`GROUP BY` is commonly used with aggregate functions such as `COUNT()`, `SUM()`, `AVG()`, `MIN()`, and `MAX()`.

### 2.5 HAVING

The **HAVING** clause is used to filter grouped data after the `GROUP BY` clause.

#### Why do We Use It?

- To filter grouped records.
- To apply conditions on aggregate functions.
- To display only the required groups.

#### Syntax

```sql
SELECT column_name, aggregate_function(column_name)
FROM table_name
GROUP BY column_name
HAVING condition;
```

#### Example

```sql
SELECT Department, COUNT(*)
FROM Employees
GROUP BY Department
HAVING COUNT(*) > 5;
```

#### Explanation

This query displays only the departments that have more than **5 employees**.

#### Note

`WHERE` filters individual rows, whereas `HAVING` filters grouped data.

### 2.6 DISTINCT

The **DISTINCT** keyword is used to return only unique values by removing duplicate records.

#### Why do We Use It?

- To eliminate duplicate values.
- To retrieve unique records.
- To improve data analysis.

#### Syntax

```sql
SELECT DISTINCT column_name
FROM table_name;
```

#### Example

```sql
SELECT DISTINCT City
FROM Students;
```

#### Explanation

If multiple students belong to the same city, the city name is displayed only once.

#### Note

`DISTINCT` can be used with one or more columns.

### 2.7 LIMIT

The **LIMIT** clause is used to restrict the number of rows returned by a query.

#### Why do We Use It?

- To retrieve a limited number of records.
- To improve query performance.
- To display sample data.

#### Syntax

```sql
SELECT column_name
FROM table_name
LIMIT number;
```

#### Example

```sql
SELECT *
FROM Students
LIMIT 5;
```

#### Explanation

This query displays only the first **5 records** from the **Students** table.

#### Note

`LIMIT` is commonly used for pagination and displaying a fixed number of rows.

# 3. Joins

A **JOIN** is used to combine rows from two or more tables based on a related column.

#### Why do We Use It?

- To combine data from multiple tables.
- To retrieve related information.
- To reduce data redundancy.
- To simplify data analysis.

The main types of joins are:

- INNER JOIN
- LEFT JOIN
- RIGHT JOIN
- FULL JOIN
- CROSS JOIN

### 3.1 INNER JOIN

The **INNER JOIN** returns only the rows that have matching values in both tables.

#### Why do We Use It?

- To retrieve matching records from two tables.
- To combine related data.
- To display only common records.

#### Syntax

```sql
SELECT columns
FROM table1
INNER JOIN table2
ON table1.column_name = table2.column_name;
```

#### Example

```sql
SELECT Students.Name, Marks.Marks
FROM Students
INNER JOIN Marks
ON Students.Student_ID = Marks.Student_ID;
```

#### Explanation

This query displays only the students who have matching records in both the **Students** and **Marks** tables.

#### Note

Only matching records from both tables are returned.

### 3.2 LEFT JOIN

The **LEFT JOIN** returns all rows from the left table and the matching rows from the right table. If there is no match, **NULL** is returned.

#### Why do We Use It?

- To retrieve all records from the left table.
- To include matching records from the right table.
- To identify missing relationships.

#### Syntax

```sql
SELECT columns
FROM table1
LEFT JOIN table2
ON table1.column_name = table2.column_name;
```

#### Example

```sql
SELECT Students.Name, Marks.Marks
FROM Students
LEFT JOIN Marks
ON Students.Student_ID = Marks.Student_ID;
```

#### Explanation

This query displays all students, even if they do not have marks.

#### Note

Non-matching rows from the right table are displayed as **NULL**.

### 3.3 RIGHT JOIN

The **RIGHT JOIN** returns all rows from the right table and the matching rows from the left table. If there is no match, **NULL** is returned.

#### Why do We Use It?

- To retrieve all records from the right table.
- To include matching records from the left table.
- To identify unmatched records.

#### Syntax

```sql
SELECT columns
FROM table1
RIGHT JOIN table2
ON table1.column_name = table2.column_name;
```

#### Example

```sql
SELECT Students.Name, Marks.Marks
FROM Students
RIGHT JOIN Marks
ON Students.Student_ID = Marks.Student_ID;
```

#### Explanation

This query displays all records from the **Marks** table, including those without matching students.

#### Note

Non-matching rows from the left table are displayed as **NULL**.

### 3.4 FULL JOIN

The **FULL JOIN** returns all rows from both tables. If there is no match, **NULL** is returned for the missing values.

#### Why do We Use It?

- To retrieve all records from both tables.
- To identify matching and non-matching records.
- To perform complete data analysis.

#### Syntax

```sql
SELECT columns
FROM table1
FULL JOIN table2
ON table1.column_name = table2.column_name;
```

#### Example

```sql
SELECT Students.Name, Marks.Marks
FROM Students
FULL JOIN Marks
ON Students.Student_ID = Marks.Student_ID;
```

#### Explanation

This query displays all records from both the **Students** and **Marks** tables.

#### Note

Matching records are combined, while non-matching records contain **NULL** values.

### 3.5 CROSS JOIN

The **CROSS JOIN** creates all possible combinations of rows from two tables. It is also called a **Cartesian Join**.

#### Why do We Use It?

- To generate all possible combinations.
- To perform combination analysis.
- To create Cartesian products.

#### Syntax

```sql
SELECT columns
FROM table1
CROSS JOIN table2;
```

#### Example

```sql
SELECT Students.Name, Marks.Marks
FROM Students
CROSS JOIN Marks;
```

#### Explanation

If there are **3 students** and **3 marks records**, the result contains **9 rows** because every student is combined with every marks record.

#### Note

`CROSS JOIN` does not require an `ON` condition because every row from the first table is combined with every row from the second table.

# 4. Advanced SQL

Advanced SQL features help perform complex queries, improve data analysis, and make database operations more efficient.

The main Advanced SQL concepts are:

- UNION
- CASE
- Subqueries
- Common Table Expressions (CTEs)
- Views
- Indexes
- Window Functions
  - ROW_NUMBER()
  - RANK()
  - DENSE_RANK()
  - LAG()
  - LEAD()

### 4.1 UNION

The **UNION** operator combines the results of two or more `SELECT` queries and removes duplicate rows.

#### Why do We Use It?

- To combine results from multiple queries.
- To remove duplicate records.
- To display data from different tables.

#### Syntax

```sql
SELECT column_name
FROM table1

UNION

SELECT column_name
FROM table2;
```

#### Example

```sql
SELECT Name
FROM Students

UNION

SELECT Name
FROM Teachers;
```

#### Explanation

This query combines the names from both **Students** and **Teachers** tables and removes duplicate names.

#### Note

Use `UNION ALL` if you want to include duplicate rows.

### 4.2 CASE

The **CASE** statement is used to apply conditions and return different values based on those conditions.

#### Why do We Use It?

- To apply conditional logic.
- To categorize data.
- To display customized results.

#### Syntax

```sql
SELECT column_name,
CASE
    WHEN condition THEN result
    ELSE result
END
FROM table_name;
```

#### Example

```sql
SELECT Name,
CASE
    WHEN Marks >= 50 THEN 'Pass'
    ELSE 'Fail'
END AS Result
FROM Students;
```

#### Explanation

This query displays **Pass** or **Fail** based on each student's marks.

#### Note

The `CASE` statement works like an **if-else** statement in programming.

### 4.3 Subqueries

A **Subquery** is a query written inside another SQL query.

#### Why do We Use It?

- To perform complex queries.
- To use the result of one query inside another query.
- To simplify data retrieval.

#### Syntax

```sql
SELECT column_name
FROM table_name
WHERE column_name = (
    SELECT function(column_name)
    FROM table_name
);
```

#### Example

```sql
SELECT Name
FROM Students
WHERE Marks = (
    SELECT MAX(Marks)
    FROM Students
);
```

#### Explanation

This query displays the student who has the highest marks.

#### Note

A subquery is also called a **Nested Query**.

### 4.4 Common Table Expressions (CTEs)

A **Common Table Expression (CTE)** is a temporary result set created using the `WITH` statement.

#### Why do We Use It?

- To simplify complex queries.
- To improve readability.
- To reuse query results.

#### Syntax

```sql
WITH CTE_Name AS (
    SELECT ...
)
SELECT *
FROM CTE_Name;
```

#### Example

```sql
WITH HighMarks AS (
    SELECT Name, Marks
    FROM Students
    WHERE Marks > 80
)
SELECT *
FROM HighMarks;
```

#### Explanation

This query creates a temporary table named **HighMarks** and displays students who scored above 80.

#### Note

A CTE exists only during the execution of the query.

### 4.5 Views

A **View** is a virtual table created from the result of a SQL query.

#### Why do We Use It?

- To simplify complex queries.
- To improve security.
- To display selected data.

#### Syntax

```sql
CREATE VIEW view_name AS
SELECT columns
FROM table_name;
```

#### Example

```sql
CREATE VIEW TopStudents AS
SELECT Name, Marks
FROM Students
WHERE Marks > 80;

SELECT * FROM TopStudents;
```

#### Explanation

This query creates a view that displays students who scored above 80.

#### Note

A view does not store data separately; it displays data from the original table.

### 4.6 Indexes

An **Index** is used to speed up data retrieval from a table.

#### Why do We Use It?

- To improve query performance.
- To search records faster.
- To reduce query execution time.

#### Syntax

```sql
CREATE INDEX index_name
ON table_name(column_name);
```

#### Example

```sql
CREATE INDEX idx_name
ON Students(Name);
```

#### Explanation

This query creates an index on the **Name** column, making searches faster.

#### Note

Indexes improve read performance but may slightly slow down insert, update, and delete operations.

### 4.7 Window Functions

Window Functions perform calculations across a set of rows without grouping them.

#### Why do We Use It?

- To rank records.
- To compare rows.
- To perform advanced data analysis.

The main Window Functions are:

- ROW_NUMBER()
- RANK()
- DENSE_RANK()
- LAG()
- LEAD()

#### 4.7.1 ROW_NUMBER()

The **ROW_NUMBER()** function assigns a unique sequential number to each row in the result.

##### Why do We Use It?

- To assign a unique number to each row.
- To arrange records in a specific order.
- To identify the position of each record.

##### Syntax

```sql
SELECT column_name,
ROW_NUMBER() OVER (ORDER BY column_name) AS Row_Num
FROM table_name;
```

##### Example

```sql
SELECT Name, Marks,
ROW_NUMBER() OVER (ORDER BY Marks DESC) AS Row_Num
FROM Students;
```

##### Explanation

This query assigns a unique row number to each student based on their marks from highest to lowest.

##### Note

`ROW_NUMBER()` always gives a unique number to each row, even when two rows have the same value.

#### 4.7.2 RANK()

The **RANK()** function assigns a rank to each row based on a specified order.

##### Why do We Use It?

- To rank records based on values.
- To identify top-performing records.
- To handle tied values with the same rank.

##### Syntax

```sql
SELECT column_name,
RANK() OVER (ORDER BY column_name) AS Rank
FROM table_name;
```

##### Example

```sql
SELECT Name, Marks,
RANK() OVER (ORDER BY Marks DESC) AS Rank
FROM Students;
```

##### Explanation

This query ranks students based on their marks from highest to lowest.

##### Note

If two students have the same marks, they receive the same rank, and the next rank is skipped.

Example:

```text
Marks    Rank
95       1
90       2
90       2
85       4
```

#### 4.7.3 DENSE_RANK()

The **DENSE_RANK()** function assigns a rank to each row without skipping ranks when there are tied values.

##### Why do We Use It?

- To rank records based on values.
- To give the same rank to tied values.
- To avoid gaps between ranks.

##### Syntax

```sql
SELECT column_name,
DENSE_RANK() OVER (ORDER BY column_name) AS Dense_Rank
FROM table_name;
```

##### Example

```sql
SELECT Name, Marks,
DENSE_RANK() OVER (ORDER BY Marks DESC) AS Dense_Rank
FROM Students;
```

##### Explanation

This query ranks students based on their marks. Students with the same marks receive the same rank.

##### Note

Unlike `RANK()`, `DENSE_RANK()` does not skip the next rank when there is a tie.

Example:

```text
Marks    Dense_Rank
95       1
90       2
90       2
85       3
```

#### 4.7.4 LAG()

The **LAG()** function returns a value from the previous row in the result.

##### Why do We Use It?

- To compare the current row with the previous row.
- To analyze changes between records.
- To access previous values without using a self-join.

##### Syntax

```sql
SELECT column_name,
LAG(column_name) OVER (ORDER BY column_name) AS Previous_Value
FROM table_name;
```

##### Example

```sql
SELECT Name, Marks,
LAG(Marks) OVER (ORDER BY Marks) AS Previous_Marks
FROM Students;
```

##### Explanation

This query displays each student's marks along with the marks from the previous row.

##### Note

The first row normally returns `NULL` because there is no previous row.

#### 4.7.5 LEAD()

The **LEAD()** function returns a value from the next row in the result.

##### Why do We Use It?

- To compare the current row with the next row.
- To analyze changes between records.
- To access the next value without using a self-join.

##### Syntax

```sql
SELECT column_name,
LEAD(column_name) OVER (ORDER BY column_name) AS Next_Value
FROM table_name;
```

##### Example

```sql
SELECT Name, Marks,
LEAD(Marks) OVER (ORDER BY Marks) AS Next_Marks
FROM Students;
```

##### Explanation

This query displays each student's marks along with the marks from the next row.

##### Note

The last row normally returns `NULL` because there is no next row.
