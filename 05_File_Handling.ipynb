## File Handling

## 1) Reading Text Files

### Definition

Reading a text file means retrieving the data stored in a file without modifying its contents.

Python uses the `open()` function in read mode (`'r'`) to read the contents of a text file.

### Why do We Read Text Files?

- To access stored data.
- To display file contents.
- To process information stored in a file.
- To avoid entering the same data repeatedly.

### Methods Used to Read a Text File

- `read()` – Reads the entire file.
- `readline()` – Reads one line from the file.
- `readlines()` – Reads all lines and returns them as a list.

### Example 1: Using `read()`

The following program reads the entire contents of a text file.

```python
file = open("sample.txt", "w")

file.write("Hello, Welcome to Python.\n")
file.write("This is a sample text file.")

file.close()

print("File created successfully.")
```

```python
file = open("sample.txt", "r")

content = file.read()
print(content)

file.close()
```

### Explanation

- `open("sample.txt", "r")` opens the file in **read** mode.
- `read()` reads the entire contents of the file.
- `print(content)` displays the file contents.
- `file.close()` closes the file.

### Example 2: Using `readline()`

The following program reads the first line of a text file.

```python
file = open("sample.txt", "r")

line = file.readline()
print(line)

file.close()
```

### Explanation

- `open("sample.txt", "r")` opens the file in **read** mode.
- `readline()` reads only one line from the file.
- `print(line)` displays the first line.
- `file.close()` closes the file.

### Example 3: Using `readlines()`

The following program reads all lines from a text file and stores them as a list.

```python
file = open("sample.txt", "r")

lines = file.readlines()
print(lines)

file.close()
```

### Explanation

- `open("sample.txt", "r")` opens the file in **read** mode.
- `readlines()` reads all lines from the file.
- Each line is stored as an element in a list.
- `print(lines)` displays the list of lines.
- `file.close()` closes the file.

### Difference Between `read()`, `readline()`, and `readlines()`

| Feature | `read()` | `readline()` | `readlines()` |
|:--------|:---------|:-------------|:--------------|
| Purpose | Reads the entire file. | Reads one line at a time. | Reads all lines from the file. |
| Return Type | `str` (String) | `str` (String) | `list` (List) |
| Number of Lines Read | All lines | One line | All lines |
| Output | Entire file as a single string. | A single line as a string. | All lines as a list of strings. |
| Memory Usage | High for large files. | Low | Moderate to High |
| Best Used When | Reading the complete file at once. | Reading one line at a time. | Processing all lines as a list. |
| Example | `file.read()` | `file.readline()` | `file.readlines()` |

## 2) Writing Text Files

### Definition

Writing a text file means storing or saving data into a file.

Python uses the `open()` function in write mode (`'w'`) to write data to a text file.

### Why do We Write Text Files?

- To store data permanently.
- To create new text files.
- To save program output.
- To update file contents.

### Methods Used to Write a Text File

- `write()` – Writes a string to the file.
- `writelines()` – Writes multiple lines from a list to the file.

### Example 1: Using `write()`

The following program writes text into a file using the `write()` method.

```python
file = open("sample.txt", "w")

file.write("Hello, Welcome to Python.\n")
file.write("This is a sample text file.")

file.close()

print("Data written successfully.")
```

### Explanation
- `open("sample.txt", "w")` opens the file in **write** mode.
- `write()` writes data to the file.
- If the file does not exist, it is created automatically.
- If the file already exists, its previous contents are overwritten.
- `file.close()` closes the file after writing.

### Example 2: Using `writelines()`

The following program writes multiple lines into a file using the `writelines()` method.

```python
file = open("sample.txt", "w")

lines = [
    "Python Programming\n",
    "File Handling\n",
    "Writing Text Files\n"
]

file.writelines(lines)

file.close()

print("Multiple lines written successfully.")
```

### Explanation
- `open("sample.txt", "w")` opens the file in **write** mode.
- `writelines()` writes multiple lines from a list to the file.
- Each item in the list is written to the file.
- `file.close()` closes the file after writing.

### Difference Between `write()` and `writelines()`

| Feature | `write()` | `writelines()` |
|----------|-----------|----------------|
| Purpose | Writes a single string to a file. | Writes multiple strings to a file. |
| Input | A string (`str`) | A list of strings (`list`) |
| Number of Lines | Writes one line or one string at a time. | Writes multiple lines in a single call. |
| Newline | Must be added manually using `\n`. | Each string should include `\n` if a new line is required. |
| Return Type | Returns the number of characters written. | Returns `None`. |
| Example | `file.write("Hello")` | `file.writelines(["Hello\n", "Python\n"])` |

## 3) CSV Files

### Definition

A CSV (Comma-Separated Values) file is a file format used to store tabular data, such as rows and columns.

Each line in a CSV file represents a row, and the values are separated by commas.

Python provides the built-in `csv` module to read from and write to CSV files.

### Why do We Use CSV Files?

- To store structured data in rows and columns.
- To exchange data between different applications.
- To save and manage large amounts of data.
- To import and export data easily.

### Common Methods

- `csv.reader()` – Reads data from a CSV file.
- `csv.writer()` – Writes data to a CSV file.


### Example 1: Using `csv.writer()`

The following program creates a CSV file and writes data into it.

```python
import csv

file = open("students.csv", "w", newline="")

writer = csv.writer(file)

writer.writerow(["ID", "Name", "Marks"])
writer.writerow([1, "John", 90])
writer.writerow([2, "Alice", 95])

file.close()

print("Data written successfully.")
```

### Explanation

- `import csv` imports the CSV module.
- `open("students.csv", "w", newline="")` opens the CSV file in write mode.
- `csv.writer(file)` creates a CSV writer object.
- `writer.writerow()` writes one row at a time.
- `file.close()` closes the file after writing.

### Example 2: Using `csv.reader()`

The following program reads data from a CSV file.

```python
import csv

file = open("students.csv", "r")

reader = csv.reader(file)

for row in reader:
    print(row)

file.close()
```

### Explanation

- `import csv` imports the CSV module.
- `open("students.csv", "r")` opens the CSV file in read mode.
- `csv.reader(file)` creates a CSV reader object.
- `for row in reader` reads and displays each row.
- `file.close()` closes the file after reading.

### Difference Between `csv.reader()` and `csv.writer()`

| Feature | `csv.reader()` | `csv.writer()` |
|:--------|:---------------|:---------------|
| Purpose | Reads data from a CSV file. | Writes data to a CSV file. |
| Mode | Read (`'r'`) | Write (`'w'`) |
| Returns | Reader object | Writer object |
| Main Method | Reads rows from a file. | Writes rows to a file. |
| Example | `csv.reader(file)` | `csv.writer(file)` |

## 4) JSON Files

### Definition

A JSON (JavaScript Object Notation) file is a lightweight file format used to store and exchange structured data.

JSON stores data as key-value pairs, making it easy for humans to read and machines to process.

Python provides the built-in `json` module to read from and write to JSON files.

### Why do We Use JSON Files?

- To store structured data.
- To exchange data between applications.
- To save configuration and application data.
- To transfer data in web applications.

### Common Methods

- `json.dump()` – Writes Python data to a JSON file.
- `json.load()` – Reads data from a JSON file.


### Example 1: Using `json.dump()`

The following program writes Python data to a JSON file.

```python
import json

student = {
    "id": 1,
    "name": "John",
    "marks": 90
}

file = open("student.json", "w")

json.dump(student, file, indent=4)

file.close()

print("Data written successfully.")
```

### Explanation

- `import json` imports Python's built-in **json** module.
- A dictionary named `student` is created.
- `open("student.json", "w")` opens the JSON file in **write** mode.
- `json.dump()` writes the dictionary to the JSON file.
- `indent=4` formats the JSON file with proper indentation.
- `file.close()` closes the file after writing.

### Example 2: Using `json.load()`

The following program reads data from a JSON file.

```python
import json

file = open("student.json", "r")

student = json.load(file)

print(student)

file.close()
```

### Explanation

- `import json` imports Python's built-in **json** module.
- `open("student.json", "r")` opens the JSON file in **read** mode.
- `json.load()` reads the JSON data and converts it into a Python dictionary.
- `print(student)` displays the dictionary.
- `file.close()` closes the file after reading.

### Difference Between `json.dump()` and `json.load()`

| Feature | `json.dump()` | `json.load()` |
|----------|---------------|---------------|
| Purpose | Writes Python data to a JSON file. | Reads data from a JSON file. |
| Operation | Write | Read |
| Input | Python object | JSON file |
| Output | JSON file | Python object (dictionary/list) |
| Example | `json.dump(data, file)` | `json.load(file)` |

## 5) Pickle Files

### Definition

A Pickle file is a binary file used to store Python objects such as lists, dictionaries, tuples, and custom objects.

Python uses the built-in `pickle` module to save and retrieve Python objects.

### Why do We Use Pickle Files?

- To save Python objects permanently.
- To store complex data structures.
- To load saved data quickly.
- To preserve the state of Python objects.

### Methods

- `pickle.dump()` – Writes a Python object to a Pickle file.
- `pickle.load()` – Reads a Python object from a Pickle file.

### Example 1: Using `pickle.dump()`

The following program stores a Python dictionary in a Pickle file.

```python
import pickle

student = {
    "id": 1,
    "name": "John",
    "marks": 90
}

file = open("student.pkl", "wb")

pickle.dump(student, file)

file.close()

print("Data written successfully.")
```

### Explanation

- `import pickle` imports Python's built-in **pickle** module.
- A dictionary named `student` is created.
- `open("student.pkl", "wb")` opens the file in **write binary** mode.
- `pickle.dump()` stores the dictionary in the Pickle file.
- `file.close()` closes the file after writing.

### Example 2: Using `pickle.load()`

The following program reads data from a Pickle file.

```python
import pickle

file = open("student.pkl", "rb")

student = pickle.load(file)

print(student)

file.close()
```

### Explanation

- `import pickle` imports Python's built-in **pickle** module.
- `open("student.pkl", "rb")` opens the file in **read binary** mode.
- `pickle.load()` reads the Pickle file and converts it back into a Python object.
- `print(student)` displays the stored dictionary.
- `file.close()` closes the file after reading.

## 6) Context Manager (`with`)

### Definition

A Context Manager is a feature in Python that automatically manages resources such as files.

The `with` statement automatically opens a file and closes it after the block of code is executed, even if an error occurs.

### Why do We Use the `with` Statement?

- Automatically closes the file.
- Reduces file-related errors.
- Makes the code shorter and cleaner.
- Improves resource management.

### Methods

- `open()` – Opens a file and requires `close()` to close it manually.
- `with open()` – Opens a file and automatically closes it after use.

### Example 1: Using `open()`

The following program reads a text file using the `open()` function.

```python
file = open("sample.txt", "r")

content = file.read()
print(content)

file.close()
```

### Explanation

- `open("sample.txt", "r")` opens the file in read mode.
- `file.read()` reads the contents of the file.
- `print(content)` displays the file contents.
- `file.close()` closes the file manually.

### Example 2: Using `with open()`

The following program reads a text file using the `with` statement.

```python
with open("sample.txt", "r") as file:
    content = file.read()
    print(content)
```

### Explanation

- `with open("sample.txt", "r")` opens the file in read mode.
- `as file` creates a file object.
- `file.read()` reads the contents of the file.
- `print(content)` displays the file contents.
- The file is closed automatically after the `with` block finishes.

### Difference Between `open()` and `with open()`

| Feature | `open()` | `with open()` |
|----------|----------|---------------|
| Closing the File | Must be closed manually using `close()`. | Closes the file automatically. |
| Resource Management | Manual | Automatic |
| Error Handling | File may remain open if an error occurs. | File is closed automatically even if an error occurs. |
| Code Length | More code | Less code |
| Example | `file = open("sample.txt", "r")` | `with open("sample.txt", "r") as file:` |

## 7) Pathlib

### Definition

The **Pathlib** module is a built-in Python module that provides an object-oriented way to work with files and directories.

It makes file and path operations simpler, more readable, and platform-independent.

### Why do We Use Pathlib?

- To work with file and directory paths easily.
- To check whether a file or folder exists.
- To create, rename, and delete files or directories.
- To write cleaner and more readable code.
- To perform file operations across different operating systems.

### Common Methods

- `Path()` – Creates a path object.
- `exists()` – Checks whether a file or folder exists.
- `mkdir()` – Creates a new directory.
- `rename()` – Renames a file or directory.
- `unlink()` – Deletes a file.

### Example 1: Using `Path()`

The following program creates a path object and displays its location.

```python
from pathlib import Path

path = Path("sample.txt")

print(path)
```

### Explanation

- `from pathlib import Path` imports the `Path` class.
- `Path("sample.txt")` creates a path object for the file.
- `print(path)` displays the file path.

### Example 2: Using `exists()`

The following program checks whether a file exists.

```python
from pathlib import Path

path = Path("sample.txt")

print(path.exists())
```

### Explanation

- `Path("sample.txt")` creates a path object.
- `exists()` checks whether the file or folder exists.
- `print()` displays `True` if the file exists; otherwise, it displays `False`.

### Difference Between `Path()` and `exists()`

| Feature | `Path()` | `exists()` |
|----------|----------|------------|
| Purpose | Creates a path object for a file or directory. | Checks whether the specified file or directory exists. |
| Return Type | `Path` object | Boolean (`True` or `False`) |
| Operation | Creates a path representation. | Verifies the existence of the path. |
| Example | `Path("sample.txt")` | `Path("sample.txt").exists()` |

## 9) Shutil Module

### Definition

The **Shutil (Shell Utilities)** module is a built-in Python module used to perform high-level file and directory operations.

It provides functions to copy, move, rename, and delete files and folders.

### Why do We Use the Shutil Module?

- To copy files and folders.
- To move files and directories.
- To rename files.
- To delete directories.
- To perform file management operations easily.

### Common Methods

- `shutil.copy()` – Copies a file.
- `shutil.move()` – Moves or renames a file or directory.


### Example 1: Using `shutil.copy()`

The following program copies one file to another location.

```python
import shutil

shutil.copy("sample.txt", "sample_copy.txt")

print("File copied successfully.")
```

### Explanation

- `import shutil` imports the built-in **shutil** module.
- `shutil.copy("sample.txt", "sample_copy.txt")` copies the contents of `sample.txt` to `sample_copy.txt`.
- If the destination file does not exist, it is created automatically.
- `print()` displays a confirmation message.

### Example 2: Using `shutil.move()`

The following program moves a file to a new location.

```python
import shutil

shutil.move("sample_copy.txt", "backup_sample.txt")

print("File moved successfully.")
```

### Explanation

- `import shutil` imports the built-in **shutil** module.
- `shutil.move()` moves or renames the specified file.
- If the destination file already exists, it may be overwritten depending on the operating system.
- `print()` displays a confirmation message.

### Difference Between `shutil.copy()` and `shutil.move()`

| Feature | `shutil.copy()` | `shutil.move()` |
|----------|-----------------|-----------------|
| Purpose | Copies a file. | Moves or renames a file. |
| Original File | Remains unchanged. | Is moved to the new location. |
| Destination | Creates a copy. | Transfers the original file. |
| Return Value | Destination file path. | Destination file path. |
| Example | `shutil.copy("a.txt", "b.txt")` | `shutil.move("a.txt", "b.txt")` |

## 10) File Validation

### Definition

File Validation is the process of checking whether a file exists, is accessible, or meets specific conditions before performing operations on it.

It helps prevent errors and ensures that file operations are performed safely.

### Why do We Use File Validation?

- To check whether a file exists.
- To avoid runtime errors.
- To verify file accessibility.
- To ensure safe file operations.

### Common Methods

- `os.path.exists()` – Checks whether a file or folder exists.
- `os.path.isfile()` – Checks whether the path is a file.
- `os.path.isdir()` – Checks whether the path is a directory.

### Example 1: Using `os.path.exists()`

The following program checks whether a file exists.

```python
import os

if os.path.exists("sample.txt"):
    print("File exists.")
else:
    print("File does not exist.")
```

### Explanation

- `import os` imports the built-in **os** module.
- `os.path.exists("sample.txt")` checks whether the specified file exists.
- If the file exists, the program displays **"File exists."**
- Otherwise, it displays **"File does not exist."**

### Example 2: Using `os.path.isfile()`

The following program checks whether the specified path is a file.

```python
import os

if os.path.isfile("sample.txt"):
    print("It is a file.")
else:
    print("It is not a file.")
```

### Explanation

- `os.path.isfile("sample.txt")` checks whether the specified path is a file.
- If it is a file, the program displays **"It is a file."**
- Otherwise, it displays **"It is not a file."**

### Difference Between `os.path.exists()` and `os.path.isfile()`

| Feature | `os.path.exists()` | `os.path.isfile()` |
|----------|--------------------|--------------------|
| Purpose | Checks whether a path exists. | Checks whether the path is a file. |
| Return Type | Boolean (`True` or `False`) | Boolean (`True` or `False`) |
| Checks | File or directory | File only |
| Example | `os.path.exists("sample.txt")` | `os.path.isfile("sample.txt")` |
