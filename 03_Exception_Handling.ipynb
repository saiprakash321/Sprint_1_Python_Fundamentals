## Exception Handling

## Introduction

### What is Exception Handling?

Exception handling is a way to handle errors in a program.

It prevents the program from crashing.

It displays meaningful error messages and allows the program to continue running.

### Why do we need Exception Handling?

Without exception handling, the program stops when an error occurs.

With exception handling, we can handle errors and continue the program safely.

## Types of Errors

### Compile-Time Error

#### Definition

A compile-time error occurs before the program starts executing.

It is usually caused by syntax mistakes in the code.

#### Examples

- Missing `:`
- Missing brackets `()`
- Incorrect indentation
- Spelling mistakes in Python keywords

---

### Runtime Error

#### Definition

A runtime error occurs while the program is running.

The syntax is correct, but an error happens during execution.

#### Examples

- Dividing a number by zero
- Accessing an invalid list index
- Opening a file that does not exist
- Using an undefined variable

## try

### Definition

The `try` block is used to write code that may cause an error.

Python executes the code inside the `try` block first.

If no error occurs, the program continues normally.

If an error occurs, Python looks for a matching `except` block.

### Example

Suppose I want to divide two numbers.

If the second number is zero, an error may occur.

I can place the risky code inside the `try` block.

```python
try:
    a = 10
    b = 0

    result = a / b
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

### Explanation

- `try:` starts the block that may produce an exception.

- `a = 10` stores the value `10` in the variable `a`.

- `b = 0` stores the value `0` in the variable `b`.

- `result = a / b` tries to divide `10` by `0`.

- Dividing a number by `0` is not allowed, so Python raises a `ZeroDivisionError`.

- `print(result)` is skipped because an exception occurs before it is executed.

- `except ZeroDivisionError:` catches the `ZeroDivisionError`.

- `print("Cannot divide by zero.")` displays a user-friendly error message instead of crashing the program.

## except

### Definition

The `except` block is used to handle errors that occur inside the `try` block.

It prevents the program from crashing and displays a meaningful message.

### Example

Suppose a user enters an invalid number.

Instead of stopping the program, I can handle the error using `except`.

```python
try:
    number = int("abc")

except ValueError:
    print("Please enter a valid number.")
```

### Explanation

- `try:` starts the block that may produce an exception.

- `number = int("abc")` tries to convert the string `"abc"` into an integer.

- `"abc"` is not a valid integer, so Python raises a `ValueError`.

- `except ValueError:` catches the `ValueError`.

- `print("Please enter a valid number.")` displays a meaningful error message instead of stopping the program.

## else

### Definition

The `else` block executes only if no exception occurs in the `try` block.

It is used to write code that should run only when the program executes successfully.

### Example

Suppose I want to divide two numbers.

If the division is successful, I can display the result using the `else` block.

```python
try:
    a = 10
    b = 2

    result = a / b

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Division successful.")
    print("Result:", result)
```

### Explanation

- `try:` starts the block that may produce an exception.

- `a = 10` stores the value `10` in the variable `a`.

- `b = 2` stores the value `2` in the variable `b`.

- `result = a / b` divides `10` by `2` and stores the result in the variable `result`.

- Since no exception occurs, the `except` block is skipped.

- `else:` executes because the `try` block runs successfully without any exception.

- `print("Division successful.")` displays a success message.

- `print("Result:", result)` prints the result of the division.

## finally

### Definition

The `finally` block always executes whether an exception occurs or not.

It is commonly used to close files, release resources, or display a final message.

### Example

Suppose I want to display a completion message after the program finishes.

I can write that code inside the `finally` block.

```python
try:
    number = 10 / 2
    print(number)

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Program completed.")
```

### Explanation

- `try:` starts the block that may produce an exception.

- `number = 10 / 2` divides `10` by `2` and stores the result in the variable `number`.

- `print(number)` displays the result of the division.

- Since no exception occurs, the `except` block is skipped.

- `finally:` always executes whether an exception occurs or not.

- `print("Program completed.")` displays the final message.

## raise

### Definition

The `raise` keyword is used to generate an exception manually.

It is useful when we want to stop the program or display a custom error message.

### Example

Suppose a student's marks cannot be negative.

If a negative mark is entered, I can raise an exception.

```python
try:
    marks = -10

    if marks < 0:
        raise ValueError("Marks cannot be negative.")

    print("Marks:", marks)

except ValueError as error:
    print(error)
```

### Explanation

- `try:` starts the block that may produce an exception.

- `marks = -10` stores the value `-10` in the variable `marks`.

- `if marks < 0:` checks whether the marks are less than `0`.

- `raise ValueError("Marks cannot be negative.")` manually raises a `ValueError` because negative marks are not allowed.

- `print("Marks:", marks)` is skipped because an exception occurs.

- `except ValueError as error:` catches the `ValueError`.

- `print(error)` displays the error message *Marks cannot be negative*

## assert

### Definition

The `assert` statement is used to check whether a condition is true.

If the condition is false, Python raises an `AssertionError`.

It is commonly used to validate data before continuing the program.

### Example

Suppose a student's attendance must be between 0 and 100.

I can use `assert` to check whether the entered attendance is valid.

```python
try:
    age = -5

    assert age >= 0, "Age cannot be negative."

    print("Valid Age")

except AssertionError as error:
    print(error)
```

### Explanation

- `try:` starts the block that may produce an exception.

- `age = -5` stores the value `-5` in the variable `age`.

- `assert age >= 0` checks whether the age is greater than or equal to `0`.

- Since the condition is false, Python raises an `AssertionError`.

- `except AssertionError as error:` catches the exception.

- `print(error)` displays the error message.

## Custom Exceptions

### Definition

A custom exception is a user-defined exception.

It is created by inheriting the built-in `Exception` class.

Custom exceptions are used to handle application-specific errors.

### Example

Suppose a password must contain at least 8 characters.

If the password is too short, I can create and raise a custom exception.

```python
class AgeError(Exception):
    pass

try:
    age = 15

    if age < 18:
        raise AgeError("Age must be 18 or above.")

except AgeError as error:
    print(error)
```

### Explanation

- `class AgeError(Exception):` creates a custom exception named `AgeError`.

- `pass` means no additional code is added to the class.

- `try:` starts the block that may produce an exception.

- `age = 15` stores the value `15` in the variable `age`.

- `if age < 18:` checks whether the age is less than `18`.

- `raise AgeError(...)` raises the custom exception.

- `except AgeError as error:` catches the custom exception.

- `print(error)` displays the error message.
