# Module 2 – Object-Oriented Programming 

## Object-Oriented Programming (OOP)

### Class

#### What is a Class?

A class is a blueprint for creating objects.

It defines the variables and methods that objects can have.

### Example

Suppose I want to represent a student.

I can create a class for it.

```python
class Student:
    pass

print(Student)
```

### Explanation

In this program, `Student` is a class.

The `class` keyword is used to create a class.

The class is ready to create objects.

### Object

#### What is an Object?

An object is an instance of a class.

It is created using the class.

### Example

Suppose I want to create a student.

I can create an object from the `Student` class.

```python
class Student:
    pass

student1 = Student()

print(student1)
```

### Explanation
In this program, `student1` is an object.

The object is created using the `Student` class.

Each object has its own identity.

### Constructor (__init__)

#### What is a Constructor?

A constructor is a special method that is automatically called when an object is created.

It is used to initialize the object's data.

We use __init__() to give initial values to object variables.

### Example

Suppose I want every student object to have a name and age when it is created.

```python
class Student:
    def __init__(self):
        self.name = "Sai"

student = Student()

print(student.name)
```

### Explanation

In this program, `Student` is a class.

`__init__()` runs automatically when the object is created.

`self.name` stores the name `"Sai"`.

`student` is an object of the `Student` class.

`print(student.name)` displays the name.

### Instance Variables

#### What are Instance Variables?

Instance variables belong to an object.

Each object has its own copy of these variables.

### Example

Suppose each student has a different name and age.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Sai", 23)
student2 = Student("Rahul", 21)

print(student1.name)
print(student2.name)
```

### Explanation

`name` and `age` are instance variables.

Each object stores its own values.

Changing one object's values does not affect another object.

### Class Variables

#### What are Class Variables?

Class variables belong to the class.

All objects share the same class variable.

### Example

Suppose all students belong to the same college.

```python
class Student:
    college = "ABC College"

student1 = Student()
student2 = Student()

print(student1.college)
print(student2.college)
```

### Explanation

`college` is a class variable.

It belongs to the class, not to individual objects.

All objects share the same value.

### Instance Methods

#### Definition

An instance method is a method that works with object data.

It uses the `self` parameter to access instance variables.

### Example

Suppose I want to display a student's name.

```python
class Student:
    def __init__(self):
        self.name = "Sai"

    def show(self):
        print(self.name)

student = Student()
student.show()
```

### Explanation

`show()` is an instance method.

It uses `self` to access the object's data.

`student.show()` calls the method.

### Static Methods

#### Definition

A static method is a method that belongs to the class.

It does not use the `self` or `cls` parameter.

### Example

Suppose I want to display a welcome message.

```python
class Student:
    @staticmethod
    def message():
        print("Welcome")

Student.message()
```

### Explanation

`message()` is a static method.

It does not use `self`.

It is called using the class name.

### Class Methods

#### Definition

A class method is a method that works with class data.

It uses the `cls` parameter to access class variables.

### Example

Suppose I want to display the college name.

```python
class Student:
    college = "ABC College"

    @classmethod
    def show(cls):
        print(cls.college)

Student.show()
```

### Explanation
We use @classmethod when we want a method to work with class variables instead of object (instance) variables.
`show()` is a class method.

It uses `cls` to access the class variable.

It is called using the class name.

### Properties (@property)

#### Definition

A property is used to access a method like a variable.

It is created using the `@property` decorator.

### Example

Suppose I want to display a student's name without calling a method.

```python
class Student:
    @property
    def name(self):
        return "Sai"

student = Student()

print(student.name)
```

### Explanation

`@property` makes the `name()` method work like a variable.

So, we write `student.name` instead of `student.name()`.

### Encapsulation

#### Definition

Encapsulation means keeping data and methods together inside a class.

It also helps protect the object's data.

### Example

Suppose I want to store a student's name inside a class.

```python
class Student:
    def __init__(self):
        self.name = "Sai"

student = Student()

print(student.name)
```

### Explanation

The `Student` class keeps the data (`name`) together.

The object's data is stored inside the class.

### Abstraction

#### Definition

Abstraction means showing only the necessary details and hiding the implementation.

It makes the program simple to use.

### Example

Suppose I press the power button on a TV.

I don't need to know how it works inside.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Bark")

dog = Dog()
dog.sound()
```

### Explanation

`Animal` is an abstract class.

`sound()` is an abstract method.

`Dog` provides the implementation of the `sound()` method.

### Inheritance

#### Definition

Inheritance allows one class to use the properties and methods of another class.

It helps to reuse code.

### Example

Suppose a child class wants to use a method from a parent class.

```python
class Parent:
    def show(self):
        print("Hello")

class Child(Parent):
    pass

child = Child()

child.show()
```

### Explanation

`Child` inherits from `Parent`.

The `Child` class can use the `show()` method without writing it again.

### Multilevel Inheritance

#### Definition

Multilevel inheritance is a type of inheritance where one class inherits from another class, and a third class inherits from the second class.

### Example

Suppose a grandson gets properties from both his father and grandfather.

```python
class Grandparent:
    def show1(self):
        print("Grandparent")

class Parent(Grandparent):
    pass

class Child(Parent):
    pass

child = Child()
child.show1()
```

### Explanation

`Parent` inherits from `Grandparent`.

`Child` inherits from `Parent`.

So, `Child` can use the `show1()` method.

### Hierarchical Inheritance

#### Definition

Hierarchical inheritance is a type of inheritance where multiple child classes inherit from the same parent class.

### Example

Suppose two children have the same parent.

```python
class Parent:
    def show(self):
        print("Hello")

class Child1(Parent):
    pass

class Child2(Parent):
    pass

child1 = Child1()
child2 = Child2()

child1.show()
child2.show()
```

### Explanation

`Child1` and `Child2` inherit from `Parent`.

Both child classes can use the `show()` method.

### Polymorphism

#### Definition

Polymorphism means "many forms".

It allows the same method name to perform different actions for different objects.

Different classes can have methods with the same name, but the output can be different.

It makes the code flexible, reusable, and easier to maintain.

### Example

Suppose different animals make different sounds.

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

### Explanation

Both classes have a method named `sound()`.

The same method name gives different outputs for different objects.

### Method Overriding

#### Definition

Method overriding means a child class creates its own version of a method that already exists in the parent class.

The child class uses the same method name as the parent class.

When the method is called, the child class method is executed instead of the parent class method.

It allows the child class to change or customize the parent's behavior.

### Example

Suppose a child wants to display its own message instead of the parent's message.

```python
class Parent:
    def show(self):
        print("Parent")

class Child(Parent):
    def show(self):
        print("Child")

child = Child()
child.show()
```

### Explanation

Both classes have a method named `show()`.

The child class replaces the parent class method.

So, the child class method is executed.

### Method Overloading (Python Approach)

#### Definition

Method overloading means creating multiple methods with the same name but different parameters.

Python does not support method overloading directly.

Instead, we can use default arguments or variable-length arguments to achieve the same result.

This allows one method to handle different numbers of arguments.

### Example

Suppose I want to greet one person or many people using the same method.

```python
class Student:
    def show(self, name="Sai"):
        print(name)

student = Student()

student.show()
student.show("Rahul")
```

### Explanation

`show()` is called with and without an argument.

If no value is given, `"Sai"` is used.

If a value is given, it is displayed instead.

This is the Python approach to method overloading.

### Composition

#### Definition

Composition means one class contains an object of another class.

One class depends on another class to perform a task.

It helps organize the code into smaller and reusable classes.

### Example

Suppose a car has an engine.

```python
class Engine:
    def start(self):
        print("Engine Started")

class Car:
    def __init__(self):
        self.engine = Engine()

car = Car()
car.engine.start()
```

### Explanation

`Car` contains an object of the `Engine` class.

The car uses the engine to start.

This is called composition.

### Aggregation

#### Definition

Aggregation means one class uses an object of another class.

Both classes can exist independently.

The object is created outside the class and then passed to it.

### Example

Suppose a student belongs to a college.

```python
class College:
    def __init__(self, name):
        self.name = name

class Student:
    def __init__(self, college):
        self.college = college

college = College("ABC College")
student = Student(college)

print(student.college.name)
```

### Explanation

`College` is created first.

The `Student` object uses the existing `College` object.

Both objects can exist separately.

This is called aggregation.

### Association

#### Definition

Association means two classes are connected or related to each other.

Both classes can work together while remaining independent.

Neither class owns the other.

### Example

Suppose a teacher teaches a student.

```python
class Teacher:
    def teach(self):
        print("Teaching")

class Student:
    pass

teacher = Teacher()
student = Student()

teacher.teach()
```

### Explanation

`Teacher` and `Student` are two separate classes.

They can work together without depending on each other.

This relationship is called association.

### Magic (Dunder) Methods

#### Definition

Magic methods are special methods in Python.

They always start and end with double underscores (`__`).

Python automatically calls these methods when needed.

They are used to give special behavior to objects.

### Example

Suppose I want to display a custom message for an object.

```python
class Student:
    def __str__(self):
        return "Student Object"

student = Student()

print(student)
```

### Explanation

`__str__()` is a magic method.

It returns a custom string for the object.

`print(student)` automatically calls `__str__()`.

### super()

#### Definition

`super()` is used to access the parent class.

It allows the child class to use the parent class methods.

It helps avoid writing the same code again.

### Example

Suppose a child wants to use the parent's method.

```python
class Parent:
    def show(self):
        print("Hello")

class Child(Parent):
    def display(self):
        super().show()

child = Child()

child.display()
```

### Explanation

`super()` calls the parent class method.

The child class uses the `show()` method without writing it again.

### __str__()

#### Definition

`__str__()` returns a readable string for an object.

It is called automatically when we use `print()`.

It makes the output easier to understand.

### Example

Suppose I want to print a student's name.

```python
class Student:
    def __str__(self):
        return "Sai"

student = Student()

print(student)
```

### Explanation

`__str__()` returns the string `"Sai"`.

`print(student)` automatically calls the `__str__()` method.

### __repr__()

#### Definition

`__repr__()` returns the official string representation of an object.

It is mainly used for debugging and development.

It should clearly describe the object.

### Example

Suppose I want to display information about a student object.

```python
class Student:
    def __repr__(self):
        return "Student('Sai')"

student = Student()

print(student)
```

### Explanation

`__repr__()` returns the official representation of the object.

`print(student)` displays the returned string.

### __len__()

#### Definition

`__len__()` returns the length of an object.

It is called automatically when we use the `len()` function.

It allows us to define the length of our own objects.

### Example

Suppose a student has three subjects.

```python
class Student:
    def __len__(self):
        return 3

student = Student()

print(len(student))
```

### Explanation

`__len__()` returns the value `3`.

`len(student)` automatically calls the `__len__()` method.

### Dataclasses (Bonus)

#### Definition

A dataclass is a special class used to store data.

It automatically creates methods like `__init__()` and `__repr__()`.

It reduces the amount of code we need to write.

### Example

Suppose I want to store a student's name and age.

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int

student = Student("Sai", 23)

print(student)
```

### Explanation

`@dataclass` creates useful methods automatically.

`Student` stores the student's name and age.

`print(student)` displays the object's data.

### Singleton Pattern (Bonus)

#### Definition

A singleton pattern allows only one object of a class to be created.

If we try to create another object, it returns the same object.

It is used when only one object is needed.

### Example

Suppose only one school principal should exist in a school.

```python
class Student:
    pass

student1 = Student()
student2 = student1

print(student1 is student2)
```

### Explanation

`student1` is an object.

`student2` refers to the same object.

So, both variables point to one object.

The output is `True`.
