# Module 4 – Logging

## What is Logging?

### Definition

Logging is the process of recording events, messages, warnings, and errors while a program is running.

It helps us understand what the program is doing and makes it easier to find and fix problems.

### Example

Suppose an application starts, processes data, and finishes successfully.

Instead of displaying everything using `print()`, we can record these events using logging.

## Why do We Use Logging?

### Definition

Logging is used to record information about a program while it is running.

It helps developers understand what the program is doing and makes it easier to find and fix problems.

### Benefits

- Helps debug programs.
- Tracks program execution.
- Records warnings and errors.
- Monitors application behavior.
- Saves information for future analysis.

## print() vs Logging

### print()

#### Definition

`print()` is used to display simple output on the screen.

It is mainly used for learning, testing, and checking program output.

The output is displayed only on the console and is not saved for future use.

---

### Logging

#### Definition

Logging is used to record program events, messages, warnings, and errors in a structured way.

It is mainly used in real-world applications to monitor program execution and troubleshoot issues.

Log messages can be displayed on the console or saved to a log file for future reference.

---

### Why is Logging Better than `print()`?

- Supports different log levels such as `DEBUG`, `INFO`, `WARNING`, `ERROR`, and `CRITICAL`.
- Saves log messages to files.
- Displays logs on the console.
- Supports custom formatting with date, time, and log level.
- Makes debugging and monitoring easier.
- Helps maintain large applications.
- Makes it easier to identify and fix errors.

---

### Main Difference

| print() | Logging |
|----------|---------|
| Displays simple output on the console. | Records messages in a structured way. |
| Mainly used for testing and learning. | Used in real-world applications. |
| Does not support log levels. | Supports different log levels. |
| Does not save output automatically. | Can save logs to a file. |
| Limited control over output. | Provides better control and formatting. |

## Log Levels

### Definition

Log levels indicate the importance or severity of log messages.

They help developers understand the type of event that occurred while the program is running.

Python provides five standard log levels.

---

### 1. DEBUG

#### Definition

`DEBUG` is used to record detailed information that helps developers debug a program.

It is mainly used during development and testing.

**Example:**
- Checking variable values.
- Tracking program execution.

---

### 2. INFO

#### Definition

`INFO` is used to record normal events that show the program is working as expected.

It provides general information about the application's execution.

**Example:**
- Application started.
- User logged in successfully.

---

### 3. WARNING

#### Definition

`WARNING` indicates that something unexpected happened, but the program can still continue running.

It alerts developers to potential issues.

**Example:**
- File is empty.
- Low disk space.

---

### 4. ERROR

#### Definition

`ERROR` indicates that a problem occurred while performing an operation.

The program may continue running, but the current operation has failed.

**Example:**
- File not found.
- Database connection failed.

---

### 5. CRITICAL

#### Definition

`CRITICAL` indicates a very serious error that may cause the program to stop.

Immediate attention is required.

**Example:**
- Application crashed.
- System failure.

## Log Formatters

### Definition

A Log Formatter defines how log messages are displayed.

It controls the format of the log message, such as the date, time, log level, and the actual message.

Using a formatter makes log files easy to read and understand.

### Why do We Use Log Formatters?

- Makes log messages clear and readable.
- Displays useful information such as date and time.
- Helps identify the log level of each message.
- Maintains a consistent format throughout the application.

### Example

A formatter can display log messages like this:

```text
2026-08-05 10:30:15 | INFO | Application started
```

````python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logging.info("Application started")
logging.warning("This is a warning")
logging.error("This is an error")
````

### Explanation

- `import logging` imports Python's built-in **logging** module.

- `logging.basicConfig()` configures the logging settings.

- `level=logging.INFO` sets the minimum log level to **INFO**, so all log messages with level **INFO**, **WARNING**, **ERROR**, and **CRITICAL** are recorded.

- `format="%(asctime)s | %(levelname)s | %(message)s"` specifies the format in which log messages are displayed.

- `%(asctime)s` displays the date and time when the log message is generated.

- `%(levelname)s` displays the severity level of the log message.

- `%(message)s` displays the actual log message.

- `logging.info("Application started")` records an information message indicating that the application has started successfully.

- `logging.warning("This is a warning")` records a warning message to indicate a potential issue.

- `logging.error("This is an error")` records an error message when a problem occurs during program execution.

- The output displays the **date and time**, **log level**, and **message** in the format specified using `logging.basicConfig()`.

## File Handler

### Definition

A File Handler is used to save log messages to a log file instead of displaying them only on the console.

It helps store logs permanently so they can be reviewed later.

### Why do We Use a File Handler?

- Saves log messages to a file.
- Helps track program execution.
- Makes debugging easier.
- Keeps a record of errors and warnings.
- Useful for future analysis.

### Example

The following program creates a log file named `application.log` and stores log messages in it.

````python
import os
import logging

# Create the logs folder if it does not exist
os.makedirs("logs", exist_ok=True)

# Configure logging
logging.basicConfig(
    filename="logs/application.log",
    level=logging.INFO,
    format="%(levelname)s - %(message)s",
    force=True
)

# Write log messages
logging.info("Application started")
logging.info("Processing data")
logging.info("Application completed")

print("Log file created successfully.")
````

### Explanation

- `import os` imports the **os** module, which is used to work with directories and files.

- `import logging` imports Python's built-in **logging** module.

- `os.makedirs("logs", exist_ok=True)` creates a folder named **logs** if it does not already exist. The `exist_ok=True` argument prevents an error if the folder already exists.

- `logging.basicConfig()` configures the logging settings for the program.

- `filename="logs/application.log"` specifies the log file where all log messages will be stored.

- `level=logging.INFO` sets the minimum log level to **INFO**, so all messages with level **INFO** and above are recorded.

- `format="%(levelname)s - %(message)s"` specifies that each log entry will display the **log level** followed by the **log message**.

- `force=True` resets any previous logging configuration and applies the new settings.

- `logging.info("Application started")` writes an information message indicating that the application has started.

- `logging.info("Processing data")` writes an information message indicating that data is being processed.

- `logging.info("Application completed")` writes an information message indicating that the application has finished successfully.

- `print("Log file created successfully.")` displays a confirmation message on the console after the log file is created.

## Console Handler

### Definition

A Console Handler is used to display log messages on the console or terminal instead of saving them to a file.

It allows developers to view log messages immediately while the program is running.

### Why do We Use a Console Handler?

- Displays log messages on the console.
- Helps monitor program execution in real time.
- Makes debugging easier.
- Does not require opening a log file.

### Example

The following program displays log messages on the console.

````python
import logging

logger = logging.getLogger("ConsoleLogger")
logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)

formatter = logging.Formatter("%(levelname)s - %(message)s")
console_handler.setFormatter(formatter)

logger.addHandler(console_handler)

logger.info("Application started")
logger.info("Processing data")
logger.info("Application completed")
````

### Explanation

- `import logging` imports Python's built-in **logging** module.

- `logging.getLogger("ConsoleLogger")` creates a logger named **ConsoleLogger**.

- `logger.setLevel(logging.INFO)` sets the minimum log level to **INFO**.

- `logging.StreamHandler()` creates a Console Handler to display log messages on the console.

- `console_handler.setLevel(logging.INFO)` records messages with level **INFO** and above.

- `logging.Formatter("%(levelname)s - %(message)s")` specifies the format of the log messages.

- `console_handler.setFormatter(formatter)` applies the formatter to the Console Handler.

- `logger.addHandler(console_handler)` connects the Console Handler to the logger.

- `logger.info("Application started")` displays an information message indicating that the application has started.

- `logger.info("Processing data")` displays an information message indicating that data is being processed.

- `logger.info("Application completed")` displays an information message indicating that the application has finished successfully.

## Rotating Logs

### Definition

Rotating Logs automatically create backup log files when the current log file becomes too large.

This helps manage log files efficiently and prevents them from becoming too large.

### Why do We Use Rotating Logs?

- Prevents large log files.
- Creates backup log files automatically.
- Saves storage space.
- Makes log management easier.

### Example

The following program creates a rotating log file.

````python
import os
import logging
from logging.handlers import RotatingFileHandler

os.makedirs("logs", exist_ok=True)

logger = logging.getLogger("RotationLogger")
logger.setLevel(logging.INFO)

logger.handlers.clear()

handler = RotatingFileHandler(
    "logs/rotating.log",
    maxBytes=200,
    backupCount=3
)

formatter = logging.Formatter("%(levelname)s - %(message)s")
handler.setFormatter(formatter)

logger.addHandler(handler)

for i in range(1, 11):
    logger.info(f"Log Message {i}")

print("Rotating log file created successfully.")
````

### Explanation

- `from logging.handlers import RotatingFileHandler` imports the Rotating File Handler.

- `os.makedirs("logs", exist_ok=True)` creates the **logs** folder if it does not exist.

- `logging.getLogger("RotationLogger")` creates a logger named **RotationLogger**.

- `logger.setLevel(logging.INFO)` sets the minimum log level to **INFO**.

- `logger.handlers.clear()` removes any existing handlers.

- `RotatingFileHandler()` creates a rotating log file.

- `maxBytes=200` sets the maximum size of the log file to **200 bytes**.

- `backupCount=3` keeps up to **3 backup log files**.

- `logging.Formatter("%(levelname)s - %(message)s")` specifies the format of log messages.

- `handler.setFormatter(formatter)` applies the formatter to the handler.

- `logger.addHandler(handler)` attaches the rotating file handler to the logger.

- `logger.info()` writes log messages to the rotating log file.

- `print()` displays a confirmation message after the log file is created.

## Logging Best Practices

### Definition

Logging Best Practices are guidelines that help developers write clear, useful, and well-organized log messages.

Following these practices makes applications easier to debug, monitor, and maintain.

### Best Practices

- Use logging instead of too many `print()` statements.
- Use the appropriate log level (`DEBUG`, `INFO`, `WARNING`, `ERROR`, or `CRITICAL`).
- Write clear and meaningful log messages.
- Use formatters to maintain a consistent log format.
- Save important log messages in log files.
- Use rotating logs for long-running applications.
- Avoid logging sensitive information such as passwords or personal data.
- Review log files regularly to identify and fix issues.
